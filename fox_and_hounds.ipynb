```python
from GameMode import GameMode
from GameBoardExecutor import GameBoardExecutor
from GameResults import GameWinnerType
```

```python
# Get user input for game mode and number of simulations
choice = int(input("Select game mode (1-4)\n1. Random fox vs Random hounds\n2. Random fox vs Minimax hounds\n3. Shortest path fox vs Random hounds\n4. Shortest path fox vs Minimax hounds\n"))
if choice < 1 or choice > 4:
    raise ValueError("Choice must be between 1 and 4")
iters = int(input("\nHow many simulations would you want to run through? (1-1000)\n"))
if iters < 1 or iters > 1000:
    raise ValueError("Choice must be between 1 and 1000")
```

```python
# Set up GameMode configurations
game_mode = GameMode(choice, iters)
```

```python
game_exec = GameBoardExecutor()
results = game_exec.run(game_mode)
```

```python
# print the results
fox_wins = 0
hounds_wins = 0
draws = 0
count = 1
display_game_specifics = (iters <= 100)
for game_result in results:
    if game_result.winner == GameWinnerType.FOX:
        fox_wins += 1
        if display_game_specifics:
            print("Game " + str(count) + ": Fox wins with " + str(game_result.turns) + " turns") 
    elif game_result.winner == GameWinnerType.HOUNDS:
        hounds_wins += 1
        if display_game_specifics:
            print("Game " + str(count) + ": Hounds win with " + str(game_result.turns) + " turns")
    elif game_result.winner == GameWinnerType.DRAW:
        draws += 1
        if display_game_specifics:
            print("Game " + str(count) + ": Draw after " + str(game_result.turns) + " turns")
    count += 1
print("Fox wins: " + str(fox_wins) + " Hounds wins: " + str(hounds_wins) + " Draws: " + str(draws))
```
